```python
import matplotlib.pyplot as plt
plt.figure(figsize=(10,6))
plt.scatter(df['Average_Salary_USD'], df['AI_Replacement_Risk'], alpha=0.5, color='purple')
plt.title('Average_Salary_USD vs AI_Replacement_Risk', fontsize=14, fontweight='bold')
plt.xlabel('Average_Salary (USD)', fontsize=12)
plt.ylabel('AI_Replacement_Risk', fontsize=12)
plt.grid(True, linestyle='--', alpha=0.6)
plt.show()
```

```python
safe_jobs = df.groupby('Job_Title')['AI_Replacement_Risk'].mean().sort_values().head(10)
print("--- AI se sabse surakshit (Safe) 10 Jobs ---")
print(safe_jobs)
```

```python
risk_jobs = df.groupby('Job_Title')['AI_Replacement_Risk'].mean().sort_values(ascending=False).head(10)
print("--- AI se sabse jyada khatre me 10 Jobs ---")
print(risk_jobs)
```

```python
import matplotlib.pyplot as plt
plt.figure(figsize=(10,6))
risk_jobs.plot(kind='bar', color='coral')
plt.title('Top 10 Jobs with Highest AI Replacement Risk', fontsize=14, fontweight='bold')
plt.xlabel('Average AI Replacement Risk Score', fontsize=12)
plt.ylabel('Job Title', fontsize=12)
plt.xticks(rotation=45, ha='right')
plt.grid(axis='x', linestyle='--', alpha=0.7)
plt.tight_layout()
plt.show()
```

```python
rich_indians = df[(df['Country'] == 'India') & (df['Average_Salary_USD'] > 100000)]
print('Total Filtered rows:', rich_indians.shape[0])
rich_indians.head()
```

```python
rich_indians.to_csv('High_Salary_India_Jobs.csv', index=False)
print('File successfully saved!')
```

```python
all_indians = df[df['Country'] =='India']
print('Total rows from India:', all_indians.shape[0])
all_indians.head()
```

```python
all_indians.to_csv('All_India_Jobs.csv', index=False)
print('Poore India ka data save ho gya hai!')
```

```python
india_top_risk = all_indians.groupby('Job_Title')['AI_Replacement_Risk'].mean().sort_values(ascending=False).head(5)
print('--- India me AI ke sabse jyaada khatre me 5 jobs ---')
print(india_top_risk)
```
