```python
import pandas as pd 
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('flipkart_com-ecommerce_sample.csv')
```

```python
df.head()
df.info()
```

```python
df.describe()
```

```python
df.isnull().sum()
```

```python
df['brand'] = df['brand'].fillna('Unknown')
 
 
```

```python
df['product_specifications'] = df['product_specifications'].fillna('Not Available')
```

```python
df.dropna(subset=['retail_price','discounted_price','image','description'], inplace=True)
```

```python
df.head()
```

```python
df.info()
 
df.isnull().sum()
```

```python
df.describe()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
plt.figure(figsize=(10, 6))
sns.histplot(df['retail_price'], bins=50, kde=True)
plt.title('Retail Price Distribution')
plt.show()
```

```python
df_filtered = df[df['retail_price'] < 10000]
plt.figure(figsize=(10, 6))
sns.histplot(df_filtered['retail_price'], bins=50, kde=True)
plt.title('Retail Price Distribution (Under 10k)')
```

```python
plt.figure(figsize=(10, 6))
sns.countplot(y='product_category', data=df, order=df['product_category'].value_counts().index[:10])
plt.title('Top 10 Product Categories')
plt.show()
```

```python
df_under_10k = df[df['retail_price'] < 10000].copy()
df_under_10k['discount_amount'] = df_under_10k['retail_price'] - df_under_10k['discounted_price']
df_under_10k['discount_percentage'] = (df_under_10k['discount_amount'] / df_under_10k['retail_price']) * 100
category_stats = df_under_10k.groupby('product_category')[['discount_amount', 'discount_percentage']].mean().sort_values(by='discount_percentage', ascending=False)
```

```python
plt.figure(figsize=(10, 6))
category_stats['discount_percentage'].head(15).plot(kind='barh', color='teal')
plt.title('Average Discount Percentage by Category (Under 10k) ')
plt.xlabel('Discount Percentage (%)')
plt.ylabel('Product Category')
plt.grid(axis='x', linestyle='--', alpha=0.7)
plt.show()
```

```python
df['profit'] = df['retail_price'] - df['discounted_price']
profit_by_cat = df.groupby('product_category')['profit'].mean().sort_values(ascending=False).head(10)
profit_by_cat.plot(kind='bar', color='green')
plt.title('Top 10 profitable Categories')
plt.show()
```

```python
# Selects only numerical columns for correlation calculation
numeric_df = df.select_dtypes(include=[np.number])
# Compute the correlation matrix
correlation_matrix = numeric_df.corr()
# Visualize using a Heatmap
plt.figure(figsize=(8,6))
sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm', fmt='.2f', linewidth=0.5)
plt.title('Correlation Heatmap: Understanding Relationship', fontsize=15, fontweight='bold')
plt.show()
# Manager Insights
print("--- Correlation between 'discount_pct' and 'rating' is positive, it meands customers are happier with higher discounts.")
print(". If the correlation between 'discounted_price' and 'rating' is negative, it might mean premium items are rated differently.")
```
